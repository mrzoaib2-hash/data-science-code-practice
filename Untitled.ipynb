```python
with open("initial.txt",encoding='utf-8') as f:
    data=f.read()
```

```python
chunks=data.split("\n\n")
print("Total chunks:", len(chunks))
```

```python
def parse_chunk(chunk):
    chunk=chunk.strip()
    sep_chunk=chunk.split('\n')
    username=sep_chunk[0]
    name=sep_chunk[1]
    no_of_posts=int(sep_chunk[2].split("post")[0].replace(",",""))
    
    no_of_followers=float(sep_chunk[3].split("follower")[0].replace(",","").replace("K","").replace("M",""))
    if ("K" in sep_chunk[3]):
        no_of_followers=int(no_of_followers*1000)
    elif ("M" in sep_chunk[3]):
         no_of_followers=int(no_of_followers*1000000)
    no_of_following = int(sep_chunk[4].split("following")[0].replace(",", ""))

    
    if len(sep_chunk)>5:
        type_of_page=sep_chunk[5]
        bio="\n".join(sep_chunk[6:])
    else:
        type_of_page="unknown"
        bio=""
    return{"username":username,"name":name,"no_of_posts":no_of_posts,"no_of_followers":no_of_followers,"no_of_following":no_of_following,"type_of_page":type_of_page,"bio":bio}
   
```

```python
all_chunks = []

for chunk in chunks:
    try:
        parsed = parse_chunk(chunk)
        all_chunks.append(parsed)
    except Exception as e:
        print("Error mila:")
        print(chunk)
        print(e)


```

```python
import json
s=json.dumps(all_chunks,indent=4)
with open("data.json","w") as f:
    f.write(s)
```
